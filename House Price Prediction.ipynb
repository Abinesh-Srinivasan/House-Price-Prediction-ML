Importing Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor
from sklearn import metrics
```

Importing Boston House Price Dataset

```python
url = "https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv"
house_price_dataset=pd.read_csv(url)
print(house_price_dataset)
```

```python
# Loading the datas into a Pandas Dataframe
house_price_dataframe=pd.DataFrame(house_price_dataset)
house_price_dataframe.head()
```

```python
#count the rows and colums
house_price_dataframe.shape
```

```python
# checking for missing data
house_price_dataframe.isnull().sum()
```

```python
# statistical computation
house_price_dataframe.describe()
```

Calculating the Correlation

```python
correlation=house_price_dataframe.corr()
```

```python
# heatmap construction
plt.figure(figsize=(10,10))
sns.heatmap(correlation,cbar=True, cmap='Reds', square=True, annot=True,fmt='.1f', annot_kws={'size':8})
```

Splitting the Feature(Input) and Target(Output)

```python
X=house_price_dataframe.drop(['medv'],axis=1)
Y=house_price_dataframe['medv']
print(X)
print(Y)
```

Splitting data for Training and Testing

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=2)
print(X.shape,X_train.shape,X_test.shape)
```

Model Training

```python
model=XGBRegressor()
model.fit(X_train,Y_train)
```

Prediction on Training data

```python
training_data_prediction=model.predict(X_train)
print(training_data_prediction)
```

Variance on Training Data Prediction

```python
# r square error
score_1=metrics.r2_score(Y_train,training_data_prediction)
# mean absolute error
score_2=metrics.mean_absolute_error(Y_train,training_data_prediction)

print(f"R Square error: {score_1}")
print(f"Mean absolute error: {score_2}")
```

Visualization on Training Data Prediction

```python
plt.scatter(Y_train,training_data_prediction)
plt.xlabel("Actual Value")
plt.ylabel("Predicted Value")
plt.title("Training Data Prediction")
plt.show()
```

Prediction on Testing data

```python
test_data_prediction=model.predict(X_test)
print(test_data_prediction)
```

Variance on Testing Data Prediction

```python
# R square error
score_3=metrics.r2_score(Y_test,test_data_prediction)
# Mean absolute error
score_4=metrics.mean_absolute_error(Y_test,test_data_prediction)

print(score_3)
print(score_4)
```

Visualization on Testing Data Prediction

```python
plt.scatter(Y_test,test_data_prediction)
plt.xlabel("Actual Value")
plt.ylabel("Predicted Value")
plt.title("Testing Data Prediction")
plt.show()
```
